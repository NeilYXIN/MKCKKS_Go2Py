# MKCKKS Go2Py: comprehensive executable tutorial

This tutorial uses a three-client federated scenario to exercise the documented public computation surface. Each party owns independent key material, evaluated ciphertexts retain the participating party IDs, and every represented party contributes a share to collective decryption. Every case prints its progress, ciphertext metadata (including tracked magnitude, error, range margin, and correctness status), decrypted result, NumPy expected value, maximum absolute error, RMS error, and a `PASS` assertion.

The presets are upstream-tested, not independently security-vetted. CKKS is approximate, so every check enforces a scale-aware primitive bound or a propagated circuit-error bound rather than a fixed tolerance, decimal rounding, or an informal noise-budget calculation. Validated input components through `2**20` do not guarantee that cascaded intermediates remain representable; the wrapper warns and permanently marks descendants when a circuit leaves the tracked numerical range. This status is engineering evidence, not a security certificate.

## What this executes

1. Real and complex encryption through both `Party` and `PublicKey`.
2. One-, two-, and three-party collective decryption.
3. Ciphertext addition/subtraction and scalar broadcasting.
4. Reusable plaintext addition, subtraction, and multiplication.
5. Real/complex scalar multiplication.
6. Ciphertext multiplication and relinearization (operation-local hoisting is an internal implementation detail).
7. Explicit rescaling and level dropping with level/scale metadata.
8. Rotation, maximum-level conjugation, and explicit blockwise multi-chunk rotation.
9. An eight-stage concatenated three-party circuit.
10. Context/CRS, private party, public key, plaintext, ciphertext, evaluation-key contribution/set, and decryption-share serialization in a restored context.

The serialization section processes large artifacts (about 108 MiB for the PN14 context and 90 MiB for the three-party evaluation-key set). Call `run(..., include_serialization=False)` for a faster computation-only pass. The implementation lives in `examples/comprehensive.py`, so the same tutorial is runnable from a terminal and testable in CI.
